#### training

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stemming_words(text):
    y = []
    for i in text.split():
        y.append(ps.stem(i))
    return " ".join(y)
new_df['tags_stemmed'] = new_df['tags'].apply(stemming_words)
```

```python
from IPython.display import display, HTML

def highlight_changes_html(x, y):
    o = x.split()
    st = y.split()
    
    result = []
    
    for i in range(min(len(o), len(st))):
        if o[i] != st[i]:
            result.append(
                f"<span style='color:red;font-weight:bold'>{o[i]} → {st[i]}</span>"
            )
        else:
            result.append(o[i])
    
    # extra words in original
    if len(o) > len(st):
        for i in range(len(st), len(o)):
            result.append(
                f"<span style='color:red;font-weight:bold'>{o[i]} → removed</span>"
            )
    
    # extra words in stemmed
    if len(st) > len(o):
        for i in range(len(o), len(st)):
            result.append(
                f"<span style='color:red;font-weight:bold'>missing → {st[i]}</span>"
            )
    
    return " ".join(result)
```

```python
new_df['highlight_diff'] = new_df.apply(
    lambda row: highlight_changes_html(row['tags'], row['tags_stemmed']),
    axis=1
)
```

```python
for i in range(5):
    display(HTML(new_df['highlight_diff'].iloc[i]))
```

```python
new_df = new_df[['id', 'title', 'tags_stemmed']]
```

```python
new_df.head(1)
```

```python
from sklearn.feature_extraction.text import CountVectorizer

cv = CountVectorizer(max_features=5000, stop_words='english')
```

```python
vectors = cv.fit_transform(new_df['tags_stemmed']).toarray()
```

```python
vectors
```

```python
from sklearn.metrics.pairwise import cosine_similarity
similar = cosine_similarity(vectors)
```

```python
similar.shape
```

```python
similar[0]
```

```python
sorted(list(enumerate(similar[0])), reverse=True, key=lambda x:x[1])
```

```python
def recommend(movie):
    movie_index = new_df[new_df['title'].str.lower() == movie.lower()].index[0]
    distances = similar[movie_index]
    movies_list = sorted(list(enumerate(distances)), reverse=True, key=lambda x : x[1])[1:6]
    
    for i in movies_list:
        print(new_df.iloc[i[0]].title)
```

```python
recommend('avatar')
```
